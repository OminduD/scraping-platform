# Scraping Platform — Data Exploration

This notebook demonstrates how to query and analyse scraped data.

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("postgresql://scraper:changeme@localhost:5432/scraping_platform")

products = pd.read_sql("SELECT * FROM products", engine)
jobs = pd.read_sql("SELECT * FROM job_listings", engine)

print(f"Products: {len(products)}, Jobs: {len(jobs)}")
```

```python
# Price distribution
products["price"].describe()
```

```python
# Products per source
products.groupby("source").size().sort_values(ascending=False)
```

```python
# Jobs by location
jobs.groupby("location").size().sort_values(ascending=False).head(10)
```
